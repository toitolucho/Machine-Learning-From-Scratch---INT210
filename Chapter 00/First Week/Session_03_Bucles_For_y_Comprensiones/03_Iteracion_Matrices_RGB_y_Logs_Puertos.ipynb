### 2. Rigor Científico: Protocolo de Iteración y Gestión de Memoria RAM

#### 2.1 El Protocolo Iterador de Python
A diferencia de los lenguajes tipo C (donde `for(int i=0; i<N; i++)` manipula un puntero numérico), en Python todo bucle `for item in coleccion:` opera sobre el **Protocolo Iterador**:
1. Invoca `iter(coleccion)` que llama internamente a `coleccion.__iter__()` para obtener un iterador.
2. En cada ciclo, ejecuta `next(iterador)` (o `iterador.__next__()`).
3. Al agotarse los elementos, el objeto lanza la excepción `StopIteration`, la cual es capturada silenciosamente por el bucle `for` para finalizar.

#### 2.2 Evaluación Perezosa (*Lazy Evaluation*) vs Listas Materializadas
- Un objeto `range(0, 100_000_000)` en Python 3 es una **secuencia virtual**: no genera cien millones de enteros en memoria, sino que almacena únicamente los parámetros `(inicio, fin, paso)` ocupando apenas **48 bytes** de memoria RAM.
- En contraste, materializar `list(range(100_000_000))` exigiría asignar aproximadamente **800 Megabytes a 3.5 Gigabytes de RAM**, pudiendo desencadenar un error `MemoryError`.

```python
import sys

# Demostración del consumo de memoria en bytes
generador_rango = range(0, 10_000_000)
lista_muestra = [i for i in range(10_000)]  # Solo 10 mil elementos

print(f"Memoria de range(10 millones de enteros): {sys.getsizeof(generador_rango)} bytes")
print(f"Memoria de lista materializada (10 mil enteros): {sys.getsizeof(lista_muestra)} bytes")

# Inspección manual del protocolo iterador
puertos_criticos = [21, 22, 23, 80]
iterador_puertos = iter(puertos_criticos)

print(f"Primer elemento obtenido con next(): {next(iterador_puertos)}")
print(f"Segundo elemento obtenido con next(): {next(iterador_puertos)}")
```

### 3. Caso de Estudio 1: Procesamiento Matricial de Píxeles RGB en Arte Digital

Representamos una imagen digital como una lista bidimensional (matriz) donde cada celda es una tupla `(R, G, B)`. Aplicamos una transformación a escala de grises mediante iteraciones anidadas y comprensiones de listas.

```python
# Matriz de 3x3 píxeles de una obra de arte digital (RGB)
imagen_rgb = [
    [(255, 0, 0),     (0, 255, 0),     (0, 0, 255)],      # Fila 0: Rojo, Verde, Azul puros
    [(255, 255, 0),   (255, 255, 255), (0, 0, 0)],        # Fila 1: Amarillo, Blanco, Negro
    [(128, 64, 200),  (200, 150, 50),  (80, 80, 80)]      # Fila 2: Colores artísticos compuestos
]

# Procesamiento píxel por píxel con enumerate para tracking de coordenadas (x, y)
print("=== Coordenadas y Valores de Luminancia ===")
for y, fila in enumerate(imagen_rgb):
    for x, (r, g, b) in enumerate(fila):
        lumi = int(0.299 * r + 0.587 * g + 0.114 * b)
        print(f"Pixel en ({x}, {y}) -> RGB: ({r:3d}, {g:3d}, {b:3d}) | Gris: {lumi:3d}")

# Transformación completa a escala de grises usando List Comprehensions
imagen_grises = [
    [int(0.299 * r + 0.587 * g + 0.114 * b) for r, g, b in fila]
    for fila in imagen_rgb
]

print("\n=== Matriz Resultante en Escala de Grises ===")
for fila_gris in imagen_grises:
    print(fila_gris)
```

### 4. Caso de Estudio 2: Auditoría de Puertos en Registros de Ciberseguridad

Analiza un log real de conexiones de red y extrae mediante comprensiones de listas únicamente aquellos accesos a **puertos vulnerables sin cifrar**.

```python
import pandas as pd

# Dataset real de registros de puertos auditados
logs_escaneo = [
    {'ip': '192.168.1.10', 'puerto': 22, 'servicio': 'SSH', 'estado': 'Abierto'},
    {'ip': '192.168.1.15', 'puerto': 23, 'servicio': 'Telnet', 'estado': 'Abierto'},
    {'ip': '10.0.0.8',     'puerto': 80, 'servicio': 'HTTP', 'estado': 'Cerrado'},
    {'ip': '172.16.1.4',   'puerto': 445, 'servicio': 'SMB', 'estado': 'Abierto'},
    {'ip': '192.168.1.99', 'puerto': 3389, 'servicio': 'RDP', 'estado': 'Abierto'},
    {'ip': '10.0.0.12',    'puerto': 443, 'servicio': 'HTTPS', 'estado': 'Abierto'}
]

PUERTOS_INSEGUROS = {23, 445, 3389}

# Filtrado declarativo pythonico mediante List Comprehension
alertas_inseguras = [
    registro for registro in logs_escaneo 
    if registro['puerto'] in PUERTOS_INSEGUROS and registro['estado'] == 'Abierto'
]

df_alertas_puertos = pd.DataFrame(alertas_inseguras)
print("=== Alertas de Puertos Inseguros Detectados ===")
display(df_alertas_puertos)
```
