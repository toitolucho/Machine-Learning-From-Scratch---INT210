### 2. Rigor Científico: Convergencia, Teorema de Punto Fijo y la Parada de Turing

#### 2.1 El Problema de la Parada de Turing (*The Halting Problem*, 1936)
Alan Turing demostró matemáticamente que no existe ningún algoritmo general que pueda determinar si un programa arbitrario finalizará o continuará ejecutándose indefinidamente. Por ello, en ingeniería de software y Machine Learning es mandatorio implementar **salvaguardas de corte temporal** (*Max Iterations Guard*) para garantizar la terminación.

#### 2.2 Convergencia Numérica y Criterio de Parada Epsilon
En los algoritmos de optimización (como el Descenso de Gradiente en Machine Learning o el Renderizado de Ray Tracing), una secuencia de aproximaciones sucesivas de estados converge hacia una solución óptima si la diferencia entre dos iteraciones consecutivas desciende por debajo de una tolerancia predefinida epsilon:

$$\text{Delta} = |x_{k+1} - x_k| < \epsilon$$

donde típicamente $\epsilon = 10^{-4}$ o $10^{-6}$.

```python
# Demostración de Convergencia Numérica hacia una tolerancia epsilon
valor_actual = 100.0
tasa_amortiguacion = 0.5
epsilon = 1e-4
iteracion = 0
max_iter = 1000

while valor_actual > epsilon and iteracion < max_iter:
    nuevo_valor = valor_actual * tasa_amortiguacion
    delta = abs(nuevo_valor - valor_actual)
    valor_actual = nuevo_valor
    iteracion += 1

print(f"Convergencia alcanzada en la iteración {iteracion} con valor residual = {valor_actual:.8f}")
```

### 3. Caso de Estudio 1: Listener de Red Continuo con Señal Centinela en Ciberseguridad

Implementa un demonio receptor de paquetes de red que procesa una cola continua de tráfico hasta interceptar un paquete de control con la señal de terminación `'FIN_ACK'`.

```python
import pandas as pd

# Flujo continuo de paquetes entrantes simulando el socket de red
cola_paquetes = [
    {'id': 1, 'flag': 'SYN',     'payload_bytes': 64,   'ip': '192.168.1.50'},
    {'id': 2, 'flag': 'ACK',     'payload_bytes': 128,  'ip': '192.168.1.50'},
    {'id': 3, 'flag': 'DATA',    'payload_bytes': 1400, 'ip': '192.168.1.50'},
    {'id': 4, 'flag': 'DATA',    'payload_bytes': 1400, 'ip': '192.168.1.50'},
    {'id': 5, 'flag': 'FIN_ACK', 'payload_bytes': 64,   'ip': '192.168.1.50'},  # Paquete Centinela
    {'id': 6, 'flag': 'DATA',    'payload_bytes': 500,  'ip': '10.0.0.99'}     # No debe procesarse
]

paquetes_procesados = []
indice = 0

# Bucle while guiado por centinela y control de flujo break
while indice < len(cola_paquetes):
    paquete_actual = cola_paquetes[indice]
    
    # Condición Centinela de parada inmediata
    if paquete_actual['flag'] == 'FIN_ACK':
        print(f"[CENTINELA DETECTADO] Sesión TCP finalizada en paquete ID #{paquete_actual['id']}.")
        paquetes_procesados.append(paquete_actual)
        break
        
    paquetes_procesados.append(paquete_actual)
    indice += 1

df_sesion = pd.DataFrame(paquetes_procesados)
print("=== Registro de la Sesión de Red Capturada ===")
display(df_sesion)
```

### 4. Caso de Estudio 2: Renderizado Adaptativo de Iluminación en Arte Digital

En la síntesis de imágenes realistas, la iluminación global se calcula iterando sobre los rebotes de luz hasta que la variación de energía entre rebotes consecutivos es insignificante (Delta < 0.001).

```python
# Algoritmo de aproximación sucesiva de iluminación ambiental
energia_inicial = 1000.0  # Lúmenes iniciales de la fuente de luz
reflectancia_material = 0.65  # El material absorbe el 35% y refleja el 65%
tolerancia_convergencia = 0.05

energia_acumulada = 0.0
rebote_actual = energia_inicial
contador_rebotes = 0
max_rebotes = 50

historial_energia = []

while rebote_actual > tolerancia_convergencia and contador_rebotes < max_rebotes:
    energia_acumulada += rebote_actual
    historial_energia.append({'rebote': contador_rebotes, 'energia_rebote': rebote_actual, 'acumulada': energia_acumulada})
    rebote_actual *= reflectancia_material
    contador_rebotes += 1

df_render = pd.DataFrame(historial_energia)
print(f"=== Renderizado Convergido en {contador_rebotes} Rebotes de Luz ===")
display(df_render.head(8))
```
