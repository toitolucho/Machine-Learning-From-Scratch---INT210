## 1. La Inteligencia Artificial, Machine Learning y Deep Learning

Para entender las diferencias, imaginemos un parque de diversiones:

* **Inteligencia Artificial (IA)** es un **Robot de Juguete con un manual**. Está programado con reglas estrictas: *"Si ves una piedra roja, salta"*. Si ve una amarilla, se confunde. No aprende, solo sigue reglas.
* **Machine Learning (ML)** es un **Cachorro Entrenable**. No tiene un manual. Le muestras 100 fotos de gatos y 100 de perros y le das un premio cada vez que acierta. Con el tiempo, aprende a reconocer patrones por sí mismo.
* **Deep Learning (DL)** es un **Detective con Supergafas**. Reconoce al gato, pero tiene gafas que desarman la imagen en millones de hilos invisibles, procesando datos súper complejos capa por capa.

## 2. Analogía de Métricas: El Tiro al Blanco

Cuando creamos un modelo que predice algo, queremos saber qué tan bueno es. Imagina que estás jugando al **Tiro al Blanco** con dardos:

* **Exactitud (Accuracy):** Es la cantidad de dardos que dieron en la diana respecto al total que lanzaste. Es como tu nota final en el colegio.
* **Precisión (Precision):** Imagina que lanzas dardos y todos caen juntos en un rinconcito del tablero. Eres muy *preciso* (consistente), aunque no hayas dado en el centro. En ML, si el modelo dice "¡Es Malware!", ¿qué tan seguro es de que no se equivocó?
* **Sensibilidad (Recall):** De todos los dardos que *debían* dar en la diana, ¿cuántos realmente diste? En Ciberseguridad, de todos los virus reales que entraron, ¿cuántos atrapó tu alarma? ¡Dejar pasar uno puede ser fatal!

## 3. Exploración Práctica con Pandas (Iris Dataset)

Hoy usaremos el dataset **Iris**, que contiene medidas de flores. Sin embargo, para nuestra mente de ingenieros, imaginemos que cada flor de la especie **Setosa** es una **Alerta de Intrusión Crítica (Malware)**, y el resto es tráfico de red legítimo de nuestro estudio de animación 3D.

```python
import pandas as pd
from sklearn.datasets import load_iris

# Carga del dataset Iris
iris_raw = load_iris()
df = pd.DataFrame(data=iris_raw.data, columns=iris_raw.feature_names)
df['species'] = iris_raw.target

# Mapeo de códigos numéricos a nombres reales de las especies
species_map = {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
df['species_name'] = df['species'].map(species_map)

# Mostrar las primeras 5 filas (Imagina que son paquetes de red analizándose)
df.head()
```

### Filtros y Consultas (Pandas Queries)

Vamos a buscar aquellas "alertas" (*setosa*) que tienen un ancho de pétalo extremadamente delgado (menor o igual a 0.2 cm), simulando un filtro de seguridad de umbral bajo.

```python
# Consulta usando query() de Pandas
filtro_setosa = df.query("species_name == 'setosa' and `petal width (cm)` <= 0.2")
print(f"Total de muestras encontradas: {len(filtro_setosa)}")
filtro_setosa.head()
```

### Análisis de Correlación Lineal

¿Crece el ancho del pétalo de forma proporcional a su longitud? Calculemos la correlación de Pearson (un valor de -1 a +1).

```python
# Cálculo de la matriz de correlación
correlacion_features = df[['petal length (cm)', 'petal width (cm)']].corr(method='pearson')
print(correlacion_features)
```

## 4. Visualización Paso a Paso (Branding Aplicado)

Vamos a graficar la relación que acabamos de encontrar, asegurándonos de aplicar la Identidad Visual de nuestro curso (Acento en Verde Azulado, Textos en Gris Pizarra).

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Aplicación de Identidad Visual
plt.rcParams.update({
    'axes.facecolor': '#F7FAFC',
    'figure.facecolor': '#F7FAFC',
    'text.color': '#4A5568',
    'axes.labelcolor': '#4A5568',
    'xtick.color': '#4A5568',
    'ytick.color': '#4A5568',
    'axes.titlecolor': '#1A365D',
})
sns.set_palette(sns.color_palette(['#319795', '#1A365D', '#4A5568']))

# 2. Creación del Gráfico base
fig, ax = plt.subplots(figsize=(8, 5))

# 3. Dibujo de los puntos
sns.scatterplot(data=df, x='petal length (cm)', y='petal width (cm)', hue='species_name', ax=ax, s=80)

# 4. Título y Etiquetas
plt.title('Relación: Longitud vs Ancho de Pétalo', fontweight='bold', fontsize=14)
plt.xlabel('Longitud del Pétalo (cm)')
plt.ylabel('Ancho del Pétalo (cm)')

plt.show()
```
